```python
import matplotlib.pyplot as plt
import csv
import numpy as np
import sklearn as sl
from sklearn import linear_model
import pandas as pd
import scipy as sp
```

```python
# Load the provided data file with the used car data (you can also have a look at it with any text editor)

filename = "data/km_year_power_price.csv"
df = pd.read_csv(filename)
df
```

### Use linear regression to estimate the car prices from the year, kilometers or engine power. 
### You can make a simple 1D regression from each one of the parameters independently 




```python
# linear regression with linregress (estimate price from year)

#x = df['powerPS'] # average price of our cars
y = df['avgPrice'] # variable that we want to relate to our x(it can be year or engine power)
#x = df['year']
#x = df['km']

xs = ['powerPS', 'year', 'km']

for x in xs:
    a = df[x]
    regression = sp.stats.linregress(a,y)
    slope = regression.slope
    intercept = regression.intercept
    rvalue = regression.rvalue

    plt.plot(a,y, '.', color = 'black', label = 'original data')
    plt.xlabel(x) # change the label
    plt.ylabel('avgPrice')
    plt.plot(a, intercept + slope*a, 'r', label = f'fitted line with R2 = {rvalue}')
    plt.legend(loc = 'best')
    plt.grid()
    plt.show()
```
